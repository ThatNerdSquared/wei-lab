```python
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import *
```

```python
fig = plt.figure(figsize=[18, 12])
x_range = np.linspace(0, 3, 100)

# haldane mapping func, with p exposed as parameter
def H(x, p):
    return (1 - np.exp(-1 * p * x)) / p

y = H(x_range, 2)
plt.plot(x_range, y)
for num in np.linspace(0.1, 2, 19):
    y = H(x_range, num)
    plt.plot(x_range, y)
plt.plot(np.linspace(0, 2.5, 100), np.linspace(0, 2.5, 100))
plt.show()
```

```python
fig = plt.figure(figsize=[18, 12])
x_range = np.linspace(0, 3, 100)

# haldane mapping func, with p exposed as parameter
def H(x, p):
    return (1 - np.exp(-1 * p * x)) / p

y = H(x_range, 2)
plt.plot(x_range, y)
plt.plot(np.linspace(0, 2, 100), np.linspace(0, 2, 100))
custom_line = plt.plot(x_range, H(x_range, 1))

def on_change(change):
    custom_line.set_ydata(H(x_range), change)
    fig.canvas.draw()

interact(on_change)
# w.observe(on_change)
plt.show()
```
