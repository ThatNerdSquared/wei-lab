define our variables as set up in the paper:

```python
x1 = var('x_1') # MAB
x2 = var('x_2') # MAb
x3 = var('x_3') # MaB
x4 = var('x_4') # Mab
x5 = var('x_5') # mAB
x6 = var('x_6') # mAb
x7 = var('x_7') # maB
x8 = var('x_8') # mab
x1, x2, x3, x4, x5, x6, x7, x8
```

kand set up parameters:

```python
R = var('R') # recomb rate between M-A loci
r1 = var('r_1') # recomb rate between A-B loci if genotype is MM
r2 = var('r_2') # ...										is Mm
r3 = var('r_3') # ...										is mm
s = var('s') # selection coeff for a or b
eps = var('epsilon') # epistasis
W1 = var('W_1') #== 1 - s # fitness of Ab or aB individual
W2 = var('W_2') #== (1 - s)^2 + eps # fitness of ab individual
Wm = var('W_m') # mean fitness
mu = var('mu') # mutation rate

R, r1, r2, r3, s, eps, W1, W2, Wm, mu
```

jkset up recursions as given in the paper / sally's mathematica code - start w chrom freqs after recomb:

```python
xr_1 = (x1-r1*(x1*x4-x2*x3)-(r2+R-2*r2*R)*(x1*x6-x2*x5)-
	R*(x1*x7-x3*x5)-(R-r2*R)*(x1*x8-x4*x5)+
	r2*R*(x3*x6-x2*x7)+r2*(x2*x7-x1*x8))
xr_2 = (x2+r1*(x1*x4-x2*x3)+(r2+R-2*r2*R)*(x1*x6-x2*x5)-
	R*(x2*x8-x4*x6)+(R-r2*R)*(x3*x6-x2*x7)-
	r2*R*(x1*x8-x4*x5)+r2*(x1*x8-x2*x7))
xr_3 = (x3+r1*(x1*x4-x2*x3)-(r2+R-2*r2*R)*(x3*x8-x4*x7)+
	R*(x1*x7-x3*x5)-(R-r2*R)*(x3*x6-x2*x7)+
	r2*R*(x1*x8-x4*x5)-r2*(x3*x6-x4*x5))
xr_4 = (x4-r1*(x1*x4-x2*x3)+(r2+R-2*r2*R)*(x3*x8-x4*x7)+
	R*(x2*x8-x4*x6)+(R-r2*R)*(x1*x8-x4*x5)-
	r2*R*(x3*x6-x2*x7)-r2*(x4*x5-x3*x6))
xr_5 = (x5-r3*(x5*x8-x6*x7)+(r2+R-2*r2*R)*(x1*x6-x2*x5)+
	R*(x1*x7-x3*x5)+(R-r2*R)*(x1*x8-x4*x5)-
	r2*R*(x3*x6-x2*x7)-r2*(x4*x5-x3*x6))
xr_6 = (x6+r3*(x5*x8-x6*x7)-(r2+R-2*r2*R)*(x1*x6-x2*x5)+
	R*(x2*x8-x4*x6)-(R-r2*R)*(x3*x6-x2*x7)+r2*R*(x1*x8-x4*x5)-
	r2*(x3*x6-x4*x5))
xr_7 = (x7+r3*(x5*x8-x6*x7)+(r2+R-2*r2*R)*(x3*x8-x4*x7)-
	R*(x1*x7-x3*x5)+(R-r2*R)*(x3*x6-x2*x7)-r2*R*(x1*x8-x4*x5)+
	r2*(x1*x8-x2*x7))
xr_8 = (x8-r3*(x5*x8-x6*x7)-(r2+R-2*r2*R)*(x3*x8-x4*x7)-
	R*(x2*x8-x4*x6)-(R-r2*R)*(x1*x8-x4*x5)+r2*R*(x3*x6-x2*x7)+
	r2*(x2*x7-x1*x8))

recursions_post_recomb = (xr_1, xr_2, xr_3, xr_4, xr_5, xr_6, xr_7, xr_8)
for eqn in recursions_post_recomb:
    show(eqn)
```

and then finally chrom freqs after mutation + selection:

```python
xp_1 = (1-mu)^2*xr_1
xp_2 = W1*(mu*(1-mu)*xr_1+(1-mu)*xr_2)
xp_3 = W1*(mu*(1-mu)*xr_1+(1-mu)*xr_3)
xp_4 = W2*(mu^2*xr_1+mu*xr_2+mu*xr_3+xr_4)
xp_5 = (1-mu)^2*xr_5
xp_6 = W1*(mu*(1-mu)*xr_5+(1-mu)*xr_6)
xp_7 = W1*(mu*(1-mu)*xr_5+(1-mu)*xr_7)
xp_8 = W2*(mu^2*xr_5+mu*xr_6+mu*xr_7+xr_8)

full_recursion_eqns = (xp_1, xp_2, xp_3, xp_4, xp_5, xp_6, xp_7, xp_8)
for eqn in full_recursion_eqns:
    show(eqn)
```

finally define mean fitness in terms of these RHSs:

```python
Wm = sum(full_recursion_eqns)
Wm
```

```python
solns = solve([Wm, 0 < s < 1, 0 < mu < 1])
len(solns)
```
